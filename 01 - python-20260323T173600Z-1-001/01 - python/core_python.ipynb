### Going Loopy

#### a. for loop

```python
mark_list = [56,78,45,90,87]

for mark in mark_list:
    if mark >= 80:
        res = 'Distinction'
    elif mark >= 60 :
        res = 'First Class'
    elif mark >= 40 :
        res = 'Second Class'
    else:
        res = 'Failed'
    print(mark, ' -> ',res)
```

```python
price_list = [75,110,35,65,25,200,30 ]
```

```python
total_price = 0
for price in price_list:
    print('Total so far:', total_price)

    print('Price of new item:', price)
    print('New Total:', total_price+price)
    print('_____________________________')

    if total_price + price > 500:
        print('Cannot afford new item.Put it back')
        break
    total_price = total_price + price
print(f'Total amt to pay ${total_price}')
```
