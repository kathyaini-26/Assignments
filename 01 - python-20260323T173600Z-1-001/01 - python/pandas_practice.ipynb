#### merge()

##### When the joining column names are SAME in both dataframes

```python
d1 = {'movieid':[1,2,3,4,5],
     'title':['3 Idiots','2 States','Leo','RRR','Pushpa']}
movies = pd.DataFrame(d1)


d2 = {
    'memid':[1,2,3,4],
    'name':['Tom','Tim','Jon','Jim'],
    'movieid':[1,np.nan,3,5]
}
members=pd.DataFrame(d2)
```

```python
movies
```

```python
members
```

```python
# Records common to botha dataframes
# Show all movies that have been rented and the members who have rented the movies
pd.merge(movies,members, on='movieid', how='inner')
```

```python
# Show ALL movies I have , if any member has rented it, show their details also
pd.merge(movies,members, on='movieid', how='left')
```

```python
# Show ALL members and the movies they have rented
pd.merge(movies,members, on='movieid', how='right')
```

```python
pd.merge(movies,members, on='movieid', how='outer')
```

##### When the joining column names are DIFFERENT in both dataframes

```python
d1 = {'id':[1,2,3,4,5],
     'title':['3 Idiots','2 States','Leo','RRR','Pushpa']}
movies2 = pd.DataFrame(d1)


d2 = {
    'memid':[1,2,3,4],
    'name':['Tom','Tim','Jon','Jim'],
    'movieid':[1,np.nan,3,5]
}
members2=pd.DataFrame(d2)
```

```python
movies2
```

```python
members2
```

```python
pd.merge(movies2,members2, left_on='id',right_on='movieid', how='inner')
```

#### Grouping Records

```python
salaries = pd.read_csv('Salaries.csv')
salaries.head()
```

```python
salaries.shape
```

```python
salaries['discipline'].nunique()
```

```python
rank_grp = salaries.groupby('rank')
rank_grp
```

```python
for group in rank_grp:
    print(group)
    print('_'*60)
```

```python
for label,records in rank_grp:
    print(label)
    print('_'*60)
```

```python
for label,records in rank_grp:
    print(records)
    print('_'*60)
```
