```python
#Data Visualization

# 3.Program bjective:

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('house_prices.csv')
print(df)

#Numerical and Categorical columns
numerical_cols=df.select_dtypes(include=['int64','float64']).columns
categorical_cols=df.select_dtypes(include=['object','category']).columns
print("Numerical Columns=",numerical_cols)
print("Categorical Columns=",categorical_cols)

# Visualize Numerical Columns
#HIstogram
df[numerical_cols].hist(figsize=(10,6),bins=15)
plt.suptitle("House Data")
plt.show()

#Boxplot
plt.figure(figsize=(10,5))
sns.boxplot(data=df[numerical_cols])
plt.title("House Data")
plt.show()

# Visualize Categorical Columns
for col in categorical_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col,data=df)
    plt.title(f"Countplot of {col}")
    plt.xticks(rotation=45)
    plt.show()

# Numerical and Categorical Relationship
for cat_col in categorical_cols:
    for num_col in numerical_cols:  
        plt.figure(figsize=(6,4))
        sns.boxplot(x=cat_col, y=num_col ,data=df)
        plt.title(f"{num_col} by {cat_col}")
        plt.xticks(rotation=45)
        plt.show()
```

```python
# 4.Program Histograms:

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

n=pd.read_csv('Cars_4vars.csv')
print(n.head())

# Numerical columns
num_cols =n.select_dtypes(include=['int64','float64']).columns
print("Numerical Columns=",num_cols)

plt.figure(figsize=(15,12))
for i,col in enumerate(num_cols, 1):
    plt.subplot(len(num_cols),1,i)
    sns.histplot(n[col].dropna(),kde=True, bins=30)
    plt.title(f"Histogram of {col}")
    plt.tight_layout()
plt.show()

# Analyze distribution
for col in num_cols:
    data=n[col].dropna()
    skew=data.skew()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = data[(data < lower) | (data > upper)]
    print(f"Column: {col}")
    print(f"  Skewness: {skew:.3f}")

    if skew > 1:
        print("  → Highly positively skewed (right-tailed).")
    elif 0.5 < skew <= 1:
        print("  → Moderately positively skewed.")
    elif -0.5 <= skew <= 0.5:
        print("  → Approximately symmetric.")
    elif -1 <= skew < -0.5:
        print("  → Moderately negatively skewed (left-tailed).")
    else:
        print("  → Highly negatively skewed (left-tailed).")

    # Outlier interpretation
    print(f"  Outliers detected: {len(outliers)}")

    if len(outliers) > 0:
        print("  → Possible extreme values present.")
    else:
        print("  → No major outliers detected.")

    print("-" * 50)
```

```python
# 5.Program Box Plots

import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('house_prices.csv')
print(df.head())

num_cols=df.select_dtypes(include=['int64','float64']).columns
print("Numerical columans=",num_cols)

for col in num_cols:
    plt.figure()
    plt.boxplot(df[col].dropna())
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.show()

for col in num_cols:
    data = df[col].dropna()
    Q1 =data.quantile(0.25)
    Q3 =data.quantile(0.75)
    IQR =Q3-Q1

    low=Q1-1.5*IQR
    upper=Q3+1.5*IQR

    outliers=data[(data <lower)|(data >upper)]

    print(f"Columns: {col}")
    print(f"IQR ={IQR:.3f}")
    print(f"Lower Bound ={lower:.3f}, Upper Bound={upper:.3f}")
    print(f"Outliers Found: {len(outliers)}","\n")

    if len(outliers)>0:
        print("Outliers detectes (possible extreme values).")
    else:
        print("No major outliers detected.")
```

```python
# Bar Chart Analysis for Categorical Column

import pandas as pd 
import matplotlib.pyplot as plt 

df= pd.read_csv("house_prices.csv")
print(df)

# Identify categorical columns

categorical_columns =df.select_dtypes(include=['object','category']).columns
print("categorical columns=")
print(list(categorical_columns))

# Create bar charts and visualize each categorical column

for col in categorical_columns:
    print(f"\n Analyzing for colum :{col}")
    count=df[col].value_counts()
    print("Category counts=")
    print(count)
    plt.figure(figsize=(8,4))
    count.plot(kind='bar',color='lightblue')
    plt.title(f"Bar chart for {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    print(f"Most common categoey in'{col}':{count.idxmax()} ({count.max()} occurrences)") 
    print("_"*40)

# 4. Conclusion Summary
print("     CONCLUSION")
print("\nKey Findings:")
for note in categorical_columns:
    print("- " + note)

print("\nOverall Summary:")
print("The bar chart analysis shows how categories are distributed across the dataset.")
print("Most frequent categories represent dominant patterns, while less frequent categories indicate rare events or values.")
print("These insights help understand the dataset structure and guide further analysis.")
```
