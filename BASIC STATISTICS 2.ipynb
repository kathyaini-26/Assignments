```python
import pandas as pd 

print("PATIENT DATA SUMMARY")
patients =pd.read_csv('patients.csv')
print(patients.info())
#print(patients)
print("\nBILLING DATA SUMMARY")
billing= pd.read_csv('billing.csv')
print(billing.info())

#Merge 
merged = pd.merge(patients, billing, on='PatientID', how='left')

# 2 Select only columns relevent for billing

billing_cols= ['PatientID', 'Department', 'Doctor', 'BillAmount']
#billing_data= patients[billing_cols]
billing_data = merged[billing_cols]
print("\n Billing related data:")
print(billing_data.head())

#3 Drop administrative columns

patients_clean = patients.drop(columns=['ReceptionistID', 'CheckInTime'],errors='ignore')
print("\n After dropping administrative columns;")
print(patients_clean)

# 4 Total bill amount per department

dep_bill = billing_data.groupby('Department')['BillAmount'].sum()
print("\n Total bill amount per department:")
print(dep_bill)

# 5 Remove duplicate patient records based on patientID

patient_unique=patients_clean.drop_duplicates(subset='PatientID')
print("\n After removing duplicate patients:")
print(patient_unique.head())

# 6 Fill missing BillAmount values with the mean

mean_bill =billing_data['BillAmount'].mean()
billing_data.loc[:,'BillAmount']=billing_data['BillAmount'].fillna(mean_bill)
print("\nAfter filling missing BillAmount values:")
print(billing_data.head())

# 7 Merge billing dataset with patient dataset on PatientID

merged= pd.merge(patients,billing,on='PatientID',how='left')
print("\nMerged dataset")
print(merged.head())

# 8 Concatenate new weekly patient data(row-wise)

new_patient= pd.DataFrame({
    'PatientID':[501, 502],
    'Name': ['New Patient A', 'New Patient B'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. X', 'Dr. Y']
})
patients_all =pd.concat([patients,new_patient],axis=0,ignore_index=True)
print("\nAfter adding new weekly patients:")
print(patients_all.head())

# 9 Concatenate new billing-related columns (column-wise)

new_columns= pd.DataFrame({
    'InsuranceCovered':[True]*len(patients_all),
    'FinalAmount':[0]*len(patients_all)})
patient_final=pd.concat([patients_all,new_columns],axis=1)
print("\n Final dataset with new billing columns:")
print(patient_final.head())
```
