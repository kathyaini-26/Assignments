```python
import math
from scipy.stats import norm

sample_mean = 3050
x_mean =600
base_cost= 1000
cost_per_unit= 5
n = 25

sigma=5*25

mu= base_cost =cost_per_unit*x_mean

standard_error= sigma/math.sqrt(n)
z= (sample_mean -mu)/standard_error

p_value =1-norm.cdf(z)

alpha=0.05
critical_value =norm.ppf(1-alpha)

print("Hypothesis Testing Results")
print(f"Theoretical Mean Cost : Rs. {mu}")
print(f"Sample Mean Cost : Rs. {sample_mean}")
print(f"Standard Deviation : {sigma}")
print(f"Sample Size (n): {n}")
print(f"Z Test Statistic: {z:.2f}")
print(f"Critical Z Value (α = 0.05): {critical_value:.2f}")
print(f"P-value: {p_value:.5f}")

print("\n")
if z> critical_value:
    print("\nDecsion: Reject the Null Hypothesis")
    print("Conclusion: There is storng evidence that weekly operating cost are higher than the model suggests,")
else:
    print("\nDecision: Fail to Reject the  Null Hypothesis")
    print("Conclusion: There is no storng evidence that weekly operating cost are higher the model suggests.")
```
